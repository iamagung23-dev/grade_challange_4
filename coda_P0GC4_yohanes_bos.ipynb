## 3. Seleksi Kolom yang Dibutuhkan

Berdasarkan key questions, kolom yang akan digunakan adalah:
- **Price** : harga properti
- **District** : provinsi/distrik
- **City** : kota
- **Type** : jenis properti
- **TotalArea** : luas total (m²)
- **LivingArea** : luas hunian (m²)
- **NumberOfBedrooms** : jumlah kamar tidur
- **NumberOfBathrooms** : jumlah kamar mandi
- **NumberOfWC** : jumlah toilet
- **ConstructionYear** : tahun bangunan
- **EnergyCertificate** : sertifikat energi
- **HasParking** : ada parkir atau tidak
- **Elevator** : ada lift atau tidak

Kolom dengan missing value lebih dari 70% di-drop karena tidak pasti untuk di analisis.

```python
kolom_terpilih = [
    'Price', 'District', 'City', 'Type',
    'TotalArea', 'LivingArea',
    'NumberOfBedrooms', 'NumberOfBathrooms', 'NumberOfWC',
    'ConstructionYear', 'EnergyCertificate',
    'HasParking', 'Elevator'
]

df_clean = df[kolom_terpilih].copy()
print('Shape setelah seleksi kolom:', df_clean.shape)
df_clean.head()
```

## 4. Data Cleaning

### 4.1 Hapus Baris Duplikat

```python
sebelum = len(df_clean)
df_clean = df_clean.drop_duplicates()
sesudah = len(df_clean)
print(f'Duplikat dihapus: {sebelum - sesudah} baris')
print(f'Sisa baris       : {sesudah}')
```

### 4.2 Hapus Baris dengan Price yang Hilang atau Tidak Wajar

Price = NaN tidak bisa dianalisis. Price < 10.000 euro kemungkinan typo/error (properti termurah Portugal realistis sekitar 20.000+ euro). Price > 50 juta euro juga outlier ekstrim (kemungkinan input salah).

```python
# Hapus baris dengan Price NaN
df_clean = df_clean.dropna(subset=['Price'])

# Filter range harga wajar: 10.000 - 50.000.000 euro
df_clean = df_clean[(df_clean['Price'] >= 10000) & (df_clean['Price'] <= 50000000)]

print('Shape setelah filter Price:', df_clean.shape)
print('\nStatistik Price setelah dibersihkan:')
print(df_clean['Price'].describe())
```

### 4.3 Handle Missing Value pada Kolom Type

Karena jumlah missing Type sangat sedikit (16 baris), di drop saja

```python
df_clean = df_clean.dropna(subset=['Type'])
print('Shape setelah handle Type:', df_clean.shape)
```

### 4.4 Imputasi Missing Value pada Kolom Numerik

Untuk kolom seperti TotalArea, LivingArea, jumlah kamar, isi missing dengan **median** per Type properti.

```python
kolom_numerik = ['TotalArea', 'LivingArea', 'NumberOfBedrooms', 
                 'NumberOfBathrooms', 'NumberOfWC', 'ConstructionYear']

for kolom in kolom_numerik:
    df_clean[kolom] = df_clean.groupby('Type')[kolom].transform(
        lambda x: x.fillna(x.median())
    )
    # Jika masih NaN (Type yang seluruhnya NaN), isi dengan median global
    df_clean[kolom] = df_clean[kolom].fillna(df_clean[kolom].median())

print('Missing value setelah imputasi:')
print(df_clean[kolom_numerik].isnull().sum())
```

### 4.5 Handle Missing Value pada Kolom Kategorik

Untuk EnergyCertificate yang missing, isi dengan 'Unknown'.

```python
df_clean['EnergyCertificate'] = df_clean['EnergyCertificate'].fillna('Unknown')
print('Distribusi EnergyCertificate:')
print(df_clean['EnergyCertificate'].value_counts())
```

### 4.6 Filter Outlier Area

Properti dengan TotalArea = 0 atau ekstrim tinggi (>10.000 m²) kemungkinannya error input.

```python
df_clean = df_clean[(df_clean['TotalArea'] > 0) & (df_clean['TotalArea'] <= 10000)]
print('Shape setelah filter TotalArea:', df_clean.shape)
```

### 4.7 Buat Kolom Turunan (Feature Engineering)

- **PricePerSqm** : harga per meter persegi
- **PropertyAge** : umur bangunan (2024 - ConstructionYear)

```python
df_clean['PricePerSqm'] = (df_clean['Price'] / df_clean['TotalArea']).round(2)
df_clean['PropertyAge'] = (2024 - df_clean['ConstructionYear']).astype(int)

# Filter umur bangunan yang masuk akal (0-200 tahun)
df_clean = df_clean[(df_clean['PropertyAge'] >= 0) & (df_clean['PropertyAge'] <= 200)]

df_clean.head()
```

## 5. Hasil Akhir dan Export

```python
print('RINGKASAN DATA CLEANING')
print(f'Jumlah baris awal   : 135,536')
print(f'Jumlah baris akhir  : {len(df_clean):,}')
print(f'Jumlah kolom akhir  : {df_clean.shape[1]}')
print(f'\nMissing values tersisa:')
print(df_clean.isnull().sum())
print(f'\nDuplikat tersisa: {df_clean.duplicated().sum()}')
```

```python
# Export ke CSV untuk diimport ke Google Spreadsheet
df_clean.to_csv('portugal_listings_cleaned.csv', index=False)
print('File berhasil disimpan: portugal_listings_cleaned.csv')
```
